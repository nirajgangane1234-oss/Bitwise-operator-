```python
#OR Operator
a =  5
b = 4
c = 12
d = 9
print(b |a)
print(c | d)
```

```python
#XOR Operator
a =  5
b = 4
c = 12
d = 9
print(a ^ b)
print(d ^ b)
```

```python
#NOT Operator
a =  5
b = 4
c = 12
d = 9
print(~a)
print(~b)
print(~c)
print(~d)
```
